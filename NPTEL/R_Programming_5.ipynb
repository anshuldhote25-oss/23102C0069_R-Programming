# LECTURE 20: NESTED if-else() AND ifelse()

```r
x = 5
if ( x==3 ) {
  x = x-1
} else if ( x < 3 ) {
  x = x+5
} else {
  x = 2*x
}
x
```

```r
x = 2
if ( x==3 ) {
  x = x-1
} else if ( x < 3 ) {
  x = x+5
} else {
  x = 2*x
}
x
```

```r
x = 3
if ( x==3 ) {
  x = x-1
} else if ( x < 3 ) {
  x = x+5
} else {
  x = 2*x
}
x
```

```r
x = 1:10
x
ifelse( x<6, x^2, x+1 )
```

```r
x = c(7, 9, 8, 4)
ifelse(x %% 2 == 0, "even number", "odd number")
```

# LECTURE 21: SWITCH() AND WHICH()

```r
switch(2, "apple", "banana", "orange")
switch(1, "apple", "banana", "orange")
```

```r
switch("colour", "colour" = "blue", "gender" = "male", "volume" = 50)
switch("volume", "colour" = "blue", "gender" = "male", "volume" = 50)
```

```r
switch(4, "apple", "banana", "orange")
switch("size", "colour" = "blue", "gender" = "male", "volume" = 50)
```

```r
x = c(10, 15, 8, 14, 6, 12)
x
which(x == 14)
which(x != 12)
which(x > 10)
```

```r
x = matrix(nrow=3, ncol=3, data=1:9)
x
which.min(x)
which.max(x)
```

```r
x = matrix(nrow=3, ncol=3, data=1:9)
x
which(x %% 2 == 1)
which(x %% 2 == 1, arr.ind = TRUE)
```
