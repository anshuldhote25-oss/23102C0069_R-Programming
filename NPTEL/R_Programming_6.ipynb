## Lecture 22 — Loops: `for` loop

### Example 1

```r
for ( i in 1:5 ) { print( i^2 ) }
```

### Example 2
`print` is a function to print the argument.

```r
for ( i in c(2,4,6,7) ) { print( i^2 ) }
```

### Example 3

```r
x = c(2,4,6,8,10,12)

excount = function(x){
  count = 0
  for (xval in x) {
    if(xval/2 > 3)
      count = count+1
  }
  print(count)
}
excount(x)
```

### Nested looping with `for` loop (Example 4)

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

### The `break` command
Using the `break` command, we can stop the loop before it has looped through all the items.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
```

### The `next` command
Using the `next` command, we can skip an iteration without terminating the loop.

Suppose we want to skip lemonade.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
```

Suppose we want to skip tea.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    next
  }
  print(x)
}
```

## Lecture 23 — Loops: `while` and `repeat`

### The `while` loop — Example 1

```r
i = 1
while (i < 10) {
  print(i^2)
  i = i+2
}
```

### The `repeat` loop — Example 3
The `repeat` loop doesn't test any condition before entering the loop, so `break` is used to terminate it.

```r
i = 1
repeat{
  print( i^2 )
  i = i+2
  if ( i > 10 )
    break
}
```

### Example 4
Additionally, the command `next` is available, to return to the beginning of the loop.

```r
i = 1
repeat{
  i = i+1
  if (i < 10) next
  print(i^2)
  if (i >= 13) break
}
```

## Lecture 24 — Functions

### Functions (Single variable)
The sign `=` or `<-` is used for defining functions.

```r
abc = function(x){
  x^2
}

abc(3)
```

```r
abc(6)
```

```r
abc(9)
```

### Functions (Two variables)

```r
abc = function(x,y){
  x^2+y^2
}

abc(3,4)
```

```r
abc(10, 10)
```

```r
abc(-2, -3)
```

### Functions — Another example

```r
abc = function(x){
  sin(x)^2+cos(x)^2 + x
}

abc(9)
```

```r
abc(99)
```

```r
abc(-15)
```

### Calling a function without an Argument

```r
abc = function() {
  for(i in 1:3) {
    print(i^3)
  }
}

abc()
```

## Lecture 25 — Sequences

The default increment is +1 or −1.

```r
seq(from=2, to=4)
```

```r
seq(from=4, to=2)
```

```r
seq(from=-4, to=4)
```

### Sequence with constant increment
Generate a sequence from 10 to 20 with an increment of 2 units.

```r
seq(from=10, to=20, by=2)
```

### Sequence with constant decrement
Generate a sequence from 20 to 10 with a decrement of 2 units.

```r
seq(from=20, to=10, by=-2)
```

### Downstream sequence with constant decrement
Generate a sequence from 3 to -2 with a decrement of 0.5 units.

```r
seq(from=3, to=-2, by=-0.5)
```

### Sequences with a predefined length, default increment +1

```r
seq(to=10, length=10)
```

```r
seq(from=10, length=10)
```

### Sequences with a predefined length, constant fractional increment

```r
seq(from=10, length=10, by=0.1)
```

### Sequences with a predefined length, constant decrement

```r
seq(from=10, length=10, by=-2)
```

### Sequences with a predefined length, constant fractional decrement

```r
seq(from=10, length=5, by=-.2)
```
