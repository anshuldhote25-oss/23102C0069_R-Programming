## Lecture 36 — `print` and `cat`

### 1. `print` only handles one object at a time

```python
%%R
print("The zero occurs at", 2*pi, "radians.")
```

### 2. The workaround: print them one at a time

```python
%%R
print("The zero occurs at"); print(2*pi); print("radians")
```

### 3. `cat` combines items into continuous output

`cat` inserts `sep = " "` between *every* argument — including `"\n"`, which is why a
space appears before the line break.

```python
%%R
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

### 4. `cat` with `date()`

```python
%%R
d = date()
cat("Today's date is:", d, "\n")
```

### 5. `cat` with a custom `sep`

```python
%%R
x = 1:10
print(x)
cat(x, sep = " ++ ")
cat("\n")   # cat does not add a newline itself
cat(x, sep = " / ")
cat("\n")
```

### 6. Mixing text and computed values

```python
%%R
x = 7
cat("The square of", x, "is", x^2, "!\n")
```

### 7. `cat` with `format()`

```python
%%R
x = 7
cat("The square root of", x, "is approximately", format(sqrt(x), digits=3), "\n")
```

### 8. `cat` prints simple vectors

```python
%%R
evenno = c(2,4,6,8,10)
print(evenno)
cat("The first few even numbers are:", evenno, "...\n")
```

### 9. `fill` and `labels`

```python
%%R
x = 1:10
cat(x, fill = 2, labels = paste("(", letters[1:10], "):"))
```

## Lecture 37 — `paste`

Unlike `cat`, the result of `paste` can be assigned to a variable.

### 10. `paste` converts arguments via `as.character`

```python
%%R
print(paste(1:12))
print(as.character(1:12))   # alternative to paste
```

### 11. The `sep` parameter

```python
%%R
print(paste("Everybody", "loves", "R Programming."))
print(paste("Everybody", "loves", "R Programming.", sep="*"))
print(paste("Everybody", "loves", "R Programming.", sep="==="))
```

### 12. Vector arguments generate all combinations

```python
%%R
names = c("Prof. Singh", "Mr.  Venkat", "Dr. Jha")
print(names)
print(paste(names, "is", "a good", "person."))
```

### 13. `collapse` — the top-level separator

```python
%%R
names = c("Prof. Singh", "Mr.  Venkat", "Dr. Jha")
paste(names, "is", "a good", "person.", collapse=", and ")
```

### 14. The role of blank space inside double quotes

```python
%%R
print(paste(1, " is first", 2, " is second", 3, " is third", sep = "#"))
print(paste(1, " is first   ", 2, " is second   ", 3, " is third   ", sep = "#"))
```

### 15. Building a name vector, then indexing it

```python
%%R
x = paste("Ex", 1:5, sep="_")
print(x)
print(x[1]); print(x[2]); print(x[3]); print(x[5])
```

### 16. `sep` vs `collapse` — the key difference

Without `collapse` you get a **vector of 5 strings**; with `collapse` you get **one string**.

```python
%%R
x = paste("Ex", 1:5, sep="_")
print(x)

x = paste("Ex", 1:5, sep="_", collapse="")
print(x)
print(x[1])
```

### 17. `paste0` vs `paste` on a single vector — identical

```python
%%R
print(paste0(1:10))
print(paste(1:10))
```

### 18. `paste0` vs `paste` on two vectors — note the blank space

```python
%%R
print(paste0(1:10, c("st", "nd", "rd", rep("th", 7))))
print(paste(1:10, c("st", "nd", "rd", rep("th", 7))))
```

## Lecture 38 — `strsplit`

`strsplit(x, split, fixed = FALSE, ...)` — `split` is treated as a **regular expression**
unless `fixed = TRUE`. It returns a **list**, one element per input string.

### 19. Same string, different separators

```python
%%R
x = "The&!syntax&!of&!paste&!is!&available!  &inthe online-help"
print(x)
print(strsplit(x, split = "!"))
print(strsplit(x, split = "&!"))
```

### 20. Accessing single components of the returned list

```python
%%R
x = "The&!syntax&!of&!paste&!is!&available!  &inthe online-help"
y = strsplit(x, split = "!&")
print(y)
print(y[[1]][1])
print(y[[1]][2])
print(y[[1]][3])   # out of range -> NA
```

### 21. Splitting dates, then converting to a matrix

```python
%%R
dates = c("2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27")
datesplt = strsplit(dates, "-")
print(datesplt)

# character matrix
datemat = matrix(unlist(datesplt), nrow = 4, ncol = 3, byrow = TRUE)
print(datemat)

# numeric matrix
datematrix = matrix(as.numeric(unlist(datesplt)), nrow = 4, ncol = 3, byrow = TRUE)
print(datematrix)
```

### 22. Splitting into individual characters with `split=""`

```python
%%R
strsplit("Shalabh", split="")
```

## Lecture 39 — `nchar`, `nzchar`, `tolower`, `toupper`

### 23. Counting characters

```python
%%R
x = "R course 24.07.2022"
y = "Number of participants: 25"
print(nchar(x))
print(nchar(y))
print(nzchar(x))
print(nzchar(y))
```

### 24. `nzchar` detects empty strings

```python
%%R
x = c("Apple", "Banana", "Cake")
print(nzchar(x))

y = c("Apple", "", "Cake")
print(y)
print(nzchar(y))
```

### 25. `nchar` on numeric vectors (coerced via `as.character` first)

```python
%%R
x = c("Apple", "Banana", "Cake"); print(nchar(x))
y = c(2, 4, 6);                   print(nchar(y))
z = c(11, 222, 3333);             print(nchar(z))
z1 = c(1.1, 2.22, 3.333);         print(nchar(z1))   # the decimal point counts
```

### 26. `tolower` and `toupper` (non-alphabetic characters unchanged)

```python
%%R
x = "R course will start from 24.07.2022"
print(toupper(x))

z = "INDIAN INSTITUTE OF TECHNOLOGY"
print(tolower(z))
```

## Quick reference

| Function | Returns | Assignable | Notes |
|---|---|---|---|
| `print(x)` | invisible `x` | — | one object at a time |
| `cat(...)` | `NULL` | no | `sep=" "` default, needs explicit `\n` |
| `paste(...)` | character vector | yes | `sep` between args, `collapse` across results |
| `paste0(...)` | character vector | yes | same as `paste(..., sep="")` |
| `strsplit(x, split)` | **list** | yes | `split` is a regex unless `fixed=TRUE` |
| `nchar(x)` | integer vector | yes | coerces to character first |
| `nzchar(x)` | logical vector | yes | `FALSE` only for `""` |
