## Lecture 16 — Logical Operators

### Reference table

| Operator | Meaning |
|---|---|
| `>` | Greater than |
| `>=` | Greater than or equal |
| `<` | Less than |
| `<=` | Less than or equal |
| `==` | Exactly equal to |
| `!=` | Not equal to |
| `!` | Negation (not) |
| `&`, `&&` | and |
| `\|`, `\|\|` | or |
| `xor(x,y)` | either x or y (exclusive or) |
| `isTRUE(x)` | test if x is TRUE |
| `isFALSE(x)` | test if x is FALSE |


### Examples: Use of `|` and `||` (scalar)

```python
%%R
x=8
(x < 10) || (x < 2)
```

```python
%%R
x=18
(x < 10) || (x < 2)
```

### Examples: Use of `|` and `||` on a vector

**Note:** The original NPTEL slides were recorded on an older R version, where `||`/`&&` on a vector silently used only the *first* element (with a warning). In current R (≥ 4.3, used in Colab today), applying `||`/`&&` to a vector of length > 1 raises an **error** instead. The cell below shows this real, current behavior using `tryCatch`, and then shows `|`, which always operates element-wise and still works exactly as in the notes.

```python
%%R
x = c(8,18)
tryCatch({
  (x < 10) || (x < 2)
}, error = function(e) print(e))
```

```python
%%R
x = c(8,18)
(x < 10) | (x < 2)   # | operates on all the elements
```

### Examples: Use of `&` and `&&` (scalar)

```python
%%R
x = 5
(x < 10) && (x > 2)
```

```python
%%R
x = 15
(x < 10) && (x > 2)
```

### Examples: Use of `&` and `&&` on a vector

Same modern-R note as above: `&&` on a vector of length > 1 errors in current R.

```python
%%R
x = c(8,18)
tryCatch({
  (x < 10) && (x > 2)
}, error = function(e) print(e))
```

```python
%%R
x = c(8,18)
(x < 10) & (x > 2)   # & operates on all the elements
```

### Examples: `&` on a range

```python
%%R
x = 1:6   # Generates x=1,2,3,4,5,6
(x > 2) & (x < 5)   # Checks whether the values are greater than 2 and less than 5
```

```python
%%R
x = 1:6
x[(x > 2) & (x < 5)]   # Finds which values are greater than 2 and smaller than 5
```

### Examples: `|` on a range

```python
%%R
x = 1:6
(x > 2) | (x < 5)   # Checks whether the values are greater than 2 or less than 5
```

```python
%%R
x = 1:6
x[(x > 2) | (x < 5)]
```

### Example: "The longer form evaluates left to right examining only the first element of each vector"

**Note:** In the original notes (older R), `(x > 2) && (x < 5)` on the vector `x = 1:6` returned `FALSE` by only checking the first element. In current R, this now raises an error (shown via `tryCatch`), because R changed `&&`/`||` to require single-length inputs for safety. The equivalent single-element comparison below still demonstrates the *intended* teaching point.

```python
%%R
x = 1:6   # Generates x = 1,2,3,4,5,6
tryCatch({
  (x > 2) && (x < 5)
}, error = function(e) print(e))
```

```python
%%R
x = 1:6
(x[1] > 2) & (x[1] < 5)   # is equivalent to checking only the first element
```

## Lecture 17 — Relational and Logical Operators

### Reference table

|---|---|
